```python
categorical_column  = ["is_patient","gender"]
quantitive_column = ['age','tot_bilirubin', 'direct_bilirubin', 'tot_proteins','albumin', 'ag_ratio', 'sgpt', 'sgot', 'alkphos']
```

```python
df.is_patient = df.is_patient.replace({
    1:1,
    2:0
})
```

```python
df.head(10)
```

```python
from sklearn.preprocessing import LabelEncoder
le= LabelEncoder()
df.gender = le.fit_transform(df.gender)
```

```python
df.head()
```

```python
X = df.drop('is_patient',axis=1)
y = df.is_patient
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42,stratify=y)
```

```python
from sklearn.preprocessing import MinMaxScaler

normalize = MinMaxScaler()

col = ['age','tot_bilirubin', 'direct_bilirubin', 'tot_proteins','albumin', 'ag_ratio', 'sgpt', 'sgot', 'alkphos']

X_train[col] = normalize.fit_transform(X_train[col])
X_test[col] = normalize.transform(X_test[col])
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier

modelk = KNeighborsClassifier(n_neighbors=5)
model = LogisticRegression(max_iter=2000, class_weight="balanced")
```

```python
model.fit(X_train,y_train)
```

```python
y_predict = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
accuracy_score(y_test,y_predict)
```

```python
modelk.fit(X_train,y_train)
```

```python
y_predictk = modelk.predict(X_test)
```

```python
accuracy_score(y_test,y_predictk)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
tr = DecisionTreeClassifier(max_depth=5,
        min_samples_leaf=5,
        class_weight="balanced",
        random_state=42)
```

```python
tr.fit(X_train,y_train)
```

```python
ypredt = tr.predict(X_test)
```

```python
accuracy_score(y_test,ypredt)
```

```python
from sklearn.ensemble import GradientBoostingClassifier
```

```python
clf =  GradientBoostingClassifier(n_estimators=100, learning_rate=1.0,
    max_depth=1, random_state=0).fit(X_train, y_train)
```

```python
yp = clf.predict(X_test)
```

```python
accuracy_score(y_test,yp)
```

```python
clf.score(X_test,y_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
ramodel =  RandomForestClassifier(n_estimators=10, max_depth=None,
    min_samples_split=2, random_state=0)
```

```python
ramodel.fit(X_test,y_test)
```

```python
pr = ramodel.predict(X_test)
```

```python
accuracy_score(y_test,pr)
```

```python
ramodel.score(X_test,y_test)
```

```python
confusion_matrix(y_test,pr)
```

```python
print(classification_report(y_test,pr))
```

```python
from sklearn.metrics  import ConfusionMatrixDisplay
print(ConfusionMatrixDisplay.from_predictions(
        y_test,
        pr,
        labels=[0, 1],
        display_labels=["Non-patient", "Patient"],
        cmap="Blues",
        colorbar=False,
    ))
```

```python
from sklearn.tree import plot_tree
```

```python
plt.figure(figsize=(20,10))
plot_tree(
    ramodel.estimators_[0],
    feature_names=X.columns,
    class_names=["NO Disease","Liver Disease"],
    filled=True,
    rounded=True
)
plt.show()
```
